```python
from pathlib import Path
import sys
here=Path.cwd().resolve()
project=next(p for p in (here,*here.parents,here/'RAG_Demo_Labs') if (p/'rag_gemini_runtime.py').exists())
sys.path.insert(0,str(project))
print('Thư mục dự án:',project)
from rag_gemini_runtime import load_index,retrieve,infer
```

```python
records=load_index('foundation')
print(f'Đã nạp {len(records)} đoạn từ index Bài 01; không tạo embedding lại.')
```

```python
print('Xây Neo4j bằng: python 01_build_neo4j_graph.py. Notebook demo dùng index Bài 01 để chat.')
```

```python
def chat(message, history):
    hits = retrieve(message, records, top_k=5)
    # Passed "Chỉ dùng bằng chứng truy xuất có trích dẫn." as extra prompt rules
    answer = infer(message, hits, "Chỉ dùng bằng chứng truy xuất có trích dẫn.")
    citations = "\n".join(
        f"- [SOURCE: {h['source']} | chunk {h['chunk_id']}]" for h in hits
    )
    return answer + "\n\n**Nguồn Hybrid/Graph RAG**\n" + citations
```

```python
import gradio as gr
demo=gr.ChatInterface(fn=chat,title='Chatbot Hybrid & Graph RAG',description='Demo truy xuất dùng index Bài 01; Neo4j dùng sau khi chạy script xây đồ thị.')
demo.launch()
```
