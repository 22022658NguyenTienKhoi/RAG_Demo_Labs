# Bài thực hành 01 — RAG Foundation

Một notebook gồm kiểm tra chunk, lập chỉ mục, truy xuất và chatbot Gradio.

```python
from pathlib import Path
import sys;
here=Path.cwd().resolve();
project=next(p for p in (here,*here.parents,here/'RAG_Demo_Labs') if (p/'rag_core').is_dir())
sys.path.insert(0,str(project))
print('Thư mục dự án:',project)
from rag_core.runtime import build_or_load_index,chunk_markdown,markdown_files,retrieve,infer
```

```python
plan={p.name:len(chunk_markdown(p)) for p in markdown_files()}
print('Kế hoạch chia đoạn:',plan,'| Tổng:',sum(plan.values()))
```

```python
records=build_or_load_index('foundation',force=False)
print(f'Đã nạp/lập chỉ mục {len(records)} đoạn.')
```

```python
def chat(message, history):
    hits = retrieve(message, records, top_k=5)
    answer = infer(message, hits)
    citations = "\n".join(
        f"- [SOURCE: {h['source']} | chunk {h['chunk_id']}]" for h in hits
    )
    return answer + "\n\n**Nguồn đã truy xuất**\n" + citations
```

```python
import gradio as gr
demo=gr.ChatInterface(fn=chat,title='Chatbot RAG cơ bản',description='Tra cứu corpus Markdown với Gemini và citation.')
demo.launch()
```
